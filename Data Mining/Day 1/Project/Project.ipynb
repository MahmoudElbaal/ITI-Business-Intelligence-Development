```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from apyori import apriori
```

```python
data = pd.read_excel('Egyptian_Grocery_Transactions.xlsx', header = None)
```

```python
data.head()
```

```python
len(data)
```

```python
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
data.columns = ['Items']
data
```

```python
items = data['Items'].values
items
```

```python
transactions = []
for i in range(0, len(items)):
    transactions.append(items[i].split(","))

transactions
```

```python
encoder = TransactionEncoder()
encoded_data = encoder.fit_transform(transactions)

encoded_data
```

```python
encodedData = pd.DataFrame(encoded_data, columns = encoder.columns_, dtype = int)
encodedData.head()
```

```python
support = apriori (encodedData, min_support = 0.1, use_colnames = True)
support.sort_values(by = 'support', ascending = False)
```

```python
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
confidence = association_rules(support, metric = 'confidence', min_threshold = 0.3)
confidence.sort_values(by = 'confidence', ascending = False)
```

```python
# Strongest Associations:
    The top rules show perfect confidence (100%):
    (سكر, بيبسي) → (بيض): When customers buy sugar and Pepsi together, they always buy eggs
    (عدس, بيبسي, بيض) → (زيت): Lentils, Pepsi and eggs always lead to oil purchase
    (بطاطس) → (فراخ): Potato purchases always lead to chicken purchases

# High-Impact Items:
    Pepsi (بيبسي) appears in most of the strongest rules
    Eggs (بيض) are frequently purchased with other items
    Oil (زيت) appears as a common consequent in high-confidence rules

# Lift Values:
    The highest lift (5.0) shows these combinations occur 5x more than random chance:
    (سكر, بيبسي) → (بيض)
    (عدس, بيبسي, بيض) → (زيت)
    (بطاطس) → (فراخ)
    
# Weak Assocciation
    Rules at the bottom show 33% confidence (1 in 3 occurrences):
    Lentils (عدس) lead to honey (عسل) 33% of the time
    Pepsi leads to eggs+oil combo 33% of the time
    
```
